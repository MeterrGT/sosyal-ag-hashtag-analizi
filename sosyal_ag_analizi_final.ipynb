```python
from pathlib import Path
import pandas as pd
input_file = Path("data/posts_hashtags_48h.csv")
if not input_file.exists():
    input_file = Path("data/posts_hashtags.csv")
df = pd.read_csv(input_file)
print("Dosya:", input_file)
print("Satır:", len(df))
print("Sütunlar:", df.columns.tolist())
display(df.head())
```

## 2. Ağ tanımı
Düğüm hashtag, kenar aynı gönderide birlikte geçen hashtag çiftidir. Kenar ağırlığı birlikte geçme sayısıdır. Kenarlar yönsüz kabul edilir.

```python
import finalize_project as fp
posts = fp.load_posts(input_file)
G = fp.graph_from_posts(posts)
print("Hashtagli gönderiler:", len(posts))
print("Düğümler:", G.number_of_nodes())
print("Kenarlar:", G.number_of_edges())
print("Yoğunluk:", round(__import__("networkx").density(G), 6))
```
